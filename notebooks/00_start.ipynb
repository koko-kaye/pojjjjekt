## 1. Czy są biblioteki?

Biblioteki z pliku `requirements.txt` instalują się same przy tworzeniu codespace. Ta komórka sprawdza, czy wszystko jest na miejscu.

```python
import importlib

biblioteki = ["pandas", "numpy", "matplotlib", "plotly", "networkx",
              "requests", "bs4", "lxml", "openai", "sklearn"]

for nazwa in biblioteki:
    try:
        modul = importlib.import_module(nazwa)
        wersja = getattr(modul, "__version__", "?")
        print(f"✅ {nazwa:12} {wersja}")
    except ImportError:
        print(f"❌ {nazwa:12} brak: uruchom komórkę poniżej")
```

Jeśli czegoś brakuje, odkomentuj i uruchom poniższą komórkę (usuń znak `#` na początku linii).

## 2. Czy jest klucz API?

Klucza **nie wpisujemy** do notatnika. Codespace czyta go z sekretu `OPENAI_API_KEY` (Settings → Secrets and variables → Codespaces). Komórka sprawdza tylko, czy sekret istnieje, i nie wyświetla jego treści.

```python
import os

klucz = os.environ.get("OPENAI_API_KEY")
if klucz:
    print(f"✅ Klucz jest dostępny (długość: {len(klucz)} znaków).")
else:
    print("❌ Brak klucza. Dodaj sekret OPENAI_API_KEY i uruchom codespace ponownie.")
```

## 3. Pierwsze zapytanie do modelu (opcjonalnie)

Uruchom tylko wtedy, gdy masz klucz. Nazwę modelu podaje prowadzący.

```python
from openai import OpenAI

MODEL = "gpt-4o-mini"  # zmień, jeśli prowadzący poda inny model

client = OpenAI()  # klucz czytany automatycznie z OPENAI_API_KEY
odpowiedz = client.chat.completions.create(
    model=MODEL,
    messages=[{"role": "user", "content": "Odpowiedz jednym zdaniem: o czym jest „Lalka” Prusa?"}],
    temperature=0,
)
print(odpowiedz.choices[0].message.content)
```
